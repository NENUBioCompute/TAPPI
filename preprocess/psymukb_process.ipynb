```python
import os
import gzip
import shutil

FILES = [f"human.{i}.rna.fna.gz" for i in range(1, 16)]  # 1~15
OUT_FASTA = "human_refseq_rna_all.fna"

def decompress_gz(in_path, out_path):
    if os.path.exists(out_path):
        print(f"Already decompressed: {out_path}")
        return
    with gzip.open(in_path, "rb") as f_in, open(out_path, "wb") as f_out:
        shutil.copyfileobj(f_in, f_out)
    print(f"Decompressed: {out_path}")

def main():
    for fname in FILES:
        decompress_gz(fname, fname[:-3])

    with open(OUT_FASTA, "wb") as out_f:
        for fname in FILES:
            unzipped = fname[:-3]
            if os.path.exists(unzipped):
                with open(unzipped, "rb") as in_f:
                    shutil.copyfileobj(in_f, out_f)
    print(f"✅ Concatenated all files into: {OUT_FASTA}")

if __name__ == "__main__":
    main()
```

```python
import os

FILES = [f"human.{i}.rna.fna" for i in range(1, 16)]

def show_first_10_records(fname):
    if not os.path.exists(fname):
        print(f"❌ File not found: {fname}")
        return
    print(f"\n===== {fname} =====")
    count = 0
    with open(fname, "r") as f:
        for line in f:
            if line.startswith(">"):
                print(line.strip())
                count += 1
                if count >= 10:
                    break
    if count == 0:
        print("(no records found)")

def main():
    for fname in FILES:
        show_first_10_records(fname)

if __name__ == "__main__":
    main()
```

```python
import os

FILES = [f"human.{i}.rna.fna" for i in range(1, 16)]

def load_fasta_to_dict(fasta_files, prefix="NM_"):
    seq_dict = {}
    for fname in fasta_files:
        if not os.path.exists(fname):
            print(f"❌ File not found: {fname}")
            continue
        with open(fname, "r") as f:
            current_id = None
            seq_lines = []
            for line in f:
                line = line.strip()
                if line.startswith(">"):
                    if current_id and seq_lines:
                        seq_dict[current_id] = "".join(seq_lines)
                    seq_lines = []
                    
                    header_id = line.split()[0][1:] 
                    if header_id.startswith(prefix):
                        current_id = header_id
                    else:
                        current_id = None
                else:
                    if current_id:
                        seq_lines.append(line)

            if current_id and seq_lines:
                seq_dict[current_id] = "".join(seq_lines)
    return seq_dict

# fasta_dict = load_fasta_to_dict(FILES)

# nm_id = "NM_001304815"
# if nm_id in fasta_dict:
#     print(f"{nm_id} sequence length: {len(fasta_dict[nm_id])}")
#     print(fasta_dict[nm_id][:100], "...") 
# else:
#     print(f"{nm_id} not found")
```
